```python
# 1. Install libraries
!pip install -q ultralytics roboflow

import os
from google.colab import files
from roboflow import Roboflow
from ultralytics import YOLO

# 2. Download your Waste-TACO dataset
rf = Roboflow(api_key="cDJekEHC6TV5etOn2TKz")
project = rf.workspace("ashleshas-workspace-nf9vb").project("waste-taco-ap2c2")
version = project.version(1)
dataset = version.download("yolov8")

# 3. Train YOLOv8 Nano (takes ~11 minutes on T4 GPU)
model = YOLO("yolov8n.pt")
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=25,
    imgsz=640,
    batch=16,
    name="wastesight_v1",
    exist_ok=True
)

# 4. Immediately download best.pt to your computer as a safe backup!
best_pt_path = "/content/runs/detect/wastesight_v1/weights/best.pt"
files.download(best_pt_path)

# 5. Export to mobile TFLite (LiteRT in Ultralytics 8.4+) and auto-download
best_model = YOLO(best_pt_path)
exported_path = best_model.export(format="litert", imgsz=640)
if exported_path and os.path.exists(str(exported_path)):
    files.download(str(exported_path))
```
